## OBJECTIVES AND EXPECTED OUTCOMES

- Predictive Risk Scores:
A machine learning model capable of generating risk scores (probabilities) that reflect a patient’s likelihood of developing lung cancer based on their profile.

- Feature Importance Analysis:
    Insights into which features (examples; smoking history, oxygen saturation levels, family history) are most predictive of lung cancer risk, guiding further research, and potential public health interventions.

- Actionable Clinical Tool:
A validated risk prediction system that can be integrated into clinical workflows, providing healthcare providers with timely, data-driven insights to support early diagnosis and personalized treatment strategies.    

## PERFORMANCE METRIC

- Accuracy:
Measures the overall percentage of correct predictions made by the model
This gives a general idea of the model’s performance but should be complemented with other metrics, especially in cases where class imbalance may exist.

- AUC-ROC (Area Under the Receiver Operating Characteristic Curve):
Evaluates the model’s ability to distinguish between the two classes (presence vs absence of pulmonary disease). A higher AUC-ROC indicates better performance across different threshold settings, which is critical for understanding the trade-off between sensitivity and specificity.


- F1 Score:
The harmonic mean of precision and recall, the F1 score is especially useful when you need a balance between false positives and false negatives. This metric is crucial in a medical context where both over-diagnosis and under-diagnosis carry significant consequences.
